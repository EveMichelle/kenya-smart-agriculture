## 📋 Project Abstract

This project builds an end-to-end machine learning platform that classifies food security risk, forecasts food prices, and analyses agricultural news sentiment across all 47 counties in Kenya — using NASA satellite weather data as the primary predictive signal.

**Business problem:** Kenya's 40%-agricultural workforce makes critical decisions with almost zero data support. FEWS NET analysts manually use NASA rainfall to assign IPC food security phases — this project automates that process with ML.

**Datasets used (all raw):**
- NASA POWER Weather API → 409,811 rows, 47 counties, 2000–2023
- FEWS NET IPC Food Security → 640 rows, 47 counties, March 2026
- KNBS CPI Monthly Reports → 5,607 text rows, 37 PDF reports, 2021–2025
- Kenya Agricultural News (scraped) → 300 articles, 2025–2026
