Documents how all 4 raw datasets were obtained. Includes fetch scripts and manual download instructions.

## 1.1 Dataset 1 — NASA POWER Weather API

**Source:** https://power.larc.nasa.gov/api/temporal/daily/point  
**How obtained:** Python API script (`kenya_nasa_power_fetch_py38.py`)  
**Coverage:** 47 counties, 2000–2023, 7 weather parameters

```python
# Load and preview NASA weather data
import pandas as pd

nasa = pd.read_csv("data/raw/weather/kenya_weather_all_counties.csv")
print("Shape:", nasa.shape)
print("Columns:", list(nasa.columns))
print("Counties:", nasa['county'].nunique())
print("\nSample:")
nasa.head(3)
```

## 1.2 Dataset 2 — FEWS NET IPC Food Security

**Source:** https://fews.net/data/acute-food-insecurity  
**How obtained:** Direct CSV download → Download All Data → Filter Kenya  
**Coverage:** 47 counties, sub-county livelihood zone level, March 2026

```python
ipc = pd.read_csv("data/raw/food_security/kenya_ipc.csv")
print("Shape:", ipc.shape)
print("Columns:", list(ipc.columns))
print("IPC phases:", ipc['ML1'].value_counts().sort_index().to_dict())
ipc.head(3)
```

## 1.3 Dataset 3 — KNBS CPI Monthly Reports

**Source:** https://www.knbs.or.ke/cpi-and-inflation-rates/  
**How obtained:** 37 PDF monthly releases → text extracted using pdfplumber → saved as CSV  
**Coverage:** 2021–2025, monthly, 20+ commodities, national level

```python
knbs = pd.read_csv("data/raw/prices/knbs_cpi_raw_text.csv")
print("Shape:", knbs.shape)
print("PDF reports:", knbs['file'].nunique())
print("\nSample content (first 200 chars):")
print(knbs['content'].iloc[0][:200])
```

## 1.4 Dataset 4 — Kenya Agricultural News (Scraped)

**Source:** https://www.kenyanews.go.ke/category/agri/  
**How obtained:** BeautifulSoup scraper (`kenya_news_scraper.py`) — 60 pages  
**Coverage:** 300 articles, 2025–2026, English language

```python
news = pd.read_csv("data/raw/news/kenya_agri_news_raw.csv")
print("Shape:", news.shape)
print("Columns:", list(news.columns))
print("Date range:", news['date'].min(), "→", news['date'].max())
news.head(3)
```
